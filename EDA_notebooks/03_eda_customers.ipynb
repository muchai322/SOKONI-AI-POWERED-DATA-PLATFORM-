```python
from pyspark.sql import functions as F

customers = spark.read.option("header", True).csv(
    "abfss://sokoni@sokoniadls.dfs.core.windows.net/bronze_layer/customers/"
)

total = customers.count()
print("rows:", total)
print("distinct customer_id:", customers.select("customer_id").distinct().count())
print("distinct email:", customers.select("email").distinct().count())

customers.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in customers.columns
]).show(vertical=True)
```
